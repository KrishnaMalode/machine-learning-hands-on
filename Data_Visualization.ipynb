```python
# to calculate upper and lower bound of boxplot
#the BLUE BOX IS CALLED IQR=interquartile range, there is 75th and 25th percentile
df2.describe()
```

```python
q3=np.percentile(df2['Age'],75)#75th percentile
q1=np.percentile(df2['Age'],25)#25th percentioe
q3,q1
```

```python
iqr=q3-q1 # box value, IQR
iqr
```

```python
#upper bound formula
ub= q3+(1.5*iqr)
lb= q1-(1.5*iqr)
ub,lb
```
