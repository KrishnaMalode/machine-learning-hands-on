## Data Manipulation using Pandas numpy

```python
import pandas as pd
```

```python
df = pd.read_csv('tested.csv')
df.head()       
```

```python
df.describe()
```

```python
df.dtypes
```

```python
## to check null values
df.isnull().sum()
```

```python
df.isnull().any(axis = 1)
```

```python
## filling missing values with mean of that column

df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

```python
df['Age'].isnull()
```

```python
## Renaming a column
df.columns
```

```python
df = df.rename(columns={'Sex':'Gender'})
df
```

```python
df.dtypes
```

```python
## Chnaging Datatype of a column
df['Age_New'] = df['Age'].astype(int)
df.dtypes
```

```python
df.head()
```

```python
## applying a function onto an entire column 

df['Age_New'] = df['Age_New'].apply(lambda x: x*2)
df
```

## Data Aggregation & Grouping

```python
df.head()
```

```python
## perform grouping based on number of people that survived based on their Gender
df.groupby(by="Gender")['Survived'].count()
```

```python
## now to group number of people that survived based on Age, Gender and Pclass
## here we can actually provide a list

df.groupby(['Gender', 'Pclass'])['Survived'].count()
```

```python
df.groupby('Survived')['Age'].mean()
```

```python
## Applying multiple aggregate functions on the same groupby function
df.groupby('Survived')['Fare'].agg(['mean','sum', 'count'])
```
