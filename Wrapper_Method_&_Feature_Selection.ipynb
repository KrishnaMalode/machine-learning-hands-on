```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('mtcars.csv')
```

```python
df.head()
```

```python
df.shape
```

```python

## seprate data
x=df[['cyl','hp','wt','disp']]
y=df['mpg']
```

```python
#Normalisation
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
```

```python
x_scaled=scaler.fit_transform(x)
```

```python
# cross vaidation
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_scaled,y,random_state=0)
```

```python
#linear regression
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x,y)
```

```python
from sklearn.metrics import r2_score
lr.score(x,y)
```

## recursive feature eliminatoin

```python
from sklearn.feature_selection import RFE
```

```python
x=df.drop('mpg',axis=1)
```

```python

rfe=RFE(lr) ## passing lr model to rfe
```

```python
rfe.fit(x,y)
```

```python
print('training acc:',rfe.score(x,y))
```

```python
## to see what features it is considering true
```

```python
rfe.get_support()
```

```python
x= x.iloc[:,rfe.get_support()]
```

```python
x
```

# backward/forward 

```python
from sklearn.feature_selection import SequentialFeatureSelector
```

```python
sfs=SequentialFeatureSelector(lr,direction='forward')
```

```python
x=df.drop('mpg',axis=1)
```

```python
sfs.fit(x,y)
```

```python
sfs.get_support()
```

```python
x= x.iloc[:,sfs.get_support()]
```

```python
x
```

```python
lr=LinearRegression()
lr.fit(x,y)
lr.score(x,y)
```
